```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
def create_dataset(num_samples=1000, img_size=64):
    X = []
    Y = []
    for _ in range(num_samples):
        img = np.zeros((img_size, img_size, 1))
        # Random circle
        rr, cc = np.random.randint(10, img_size-10), np.random.randint(10, img_size-10)
        radius = np.random.randint(5, 15)
        y, x = np.ogrid[:img_size, :img_size]
        mask = (x-cc)**2 + (y-rr)**2 <= radius**2
        img[mask] = 1.0
        X.append(img)
        Y.append(np.repeat(img, 3, axis=-1))  # target is colored version
    return np.array(X), np.array(Y)

X, Y = create_dataset(200)
```

```python
generator = tf.keras.Sequential([
    tf.keras.layers.Conv2D(64, 4, strides=2, padding='same', activation='relu', input_shape=(64,64,1)),
    tf.keras.layers.Conv2DTranspose(3, 4, strides=2, padding='same', activation='tanh')
])
```

```python
discriminator = tf.keras.Sequential([
    tf.keras.layers.Conv2D(64, 4, strides=2, padding='same', activation='relu', input_shape=(64,64,3)),
    tf.keras.layers.Conv2D(128, 4, strides=2, padding='same', activation='relu'),
    tf.keras.layers.Conv2D(1, 4, strides=1, padding='same')
])
```

```python
generator.compile(optimizer='adam', loss='binary_crossentropy')
discriminator.compile(optimizer='adam', loss='binary_crossentropy')
```
